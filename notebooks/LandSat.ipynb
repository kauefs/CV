## GeoSpatial Data Abstraction Library

GeoSpatial Data Abstraction Library **(**[GDAL](https://gdal.org/api/python/osgeo.gdal.html)**):** an invaluable tool for working with raster **&** vector geospatial data, providing tools that allow for efficient programming **&** manipulation of that type of data.

Raster Band: an important concept in geospatial image manipulation. Common color images typically have three raster bands, which represent the intensity components of blue, green, and red light. However, a raster band can contain any set of values that represent the same spatial area, not limiting itself to only light intensity information.

The extraction of raster data using GDAL is done one band at a time. The `GetRasterBand( )` function allows for the extraction of a single raster band from a GeoTIFF image.

## Normalized Difference Vegetation Index (NDVI)

Index used to demonstrate the difference between visible **&** near-infrared reflectance of vegetation cover, which **can be used to estimate the density of greenness in an area**, calculated according to the following formula:

$$
NDVI=\frac{(NIR-RED)}{(NIR+RED)}
$$

RED **&** NIR represent spectral reflectance measurements acquired in the red (visible) **&** near-infrared regions, respectively. The NDVI for an area with vegetation typically ranges from 0.3 (light vegetation) to 0.8 (dense vegetation). Clouds have a negative index, while soil and water bodies have low positive values. This information is used to estimate vegetation in a given region.

In normalized values (i.e., between 0 **&** 255), the interval of interest corresponds to intensities between 165 **&** 230.

**Near-InfraRed Region (NIR)**

NIR has a wavelength between 0.76μm **&** 0.90μm. Vegetation absorbs visible light between 0.4μm (blue light) **&** 0.7μm (red light) for photosynthesis, with maximum absorption at both ends of this spectrum. Plants appear green because they typically do not absorb light in the green wavelengths (≈0.5μm) in the middle of that range. The green band, however, is less accurate for vegetation detection because it falls within the full range of wavelengths used for photosynthesis. Conversely, plants strongly reflect near-infrared light, from 0.7μm to 1.1μm, which is exploited in NDVI calculation.

````python
# @title Functions
# Extracting Images:
def CloudMask(image):
    qa = image.select('QA_PIXEL')
    # Bits 3 & 5 are cloud shadow & cloud, respectively:
    cloudShadowBitMask = 1 << 3
    cloudsBitMask      = 1 << 5
    # Both flags should be set to zero, indicating clear conditions:
    mask = qa.bitwiseAnd(cloudShadowBitMask).eq(0).And(qa.bitwiseAnd(cloudsBitMask).eq(0))
    return image.updateMask(mask)
# Exhibiting Sattelite Images in Maps:
def AddLayer(self, img, params, name):
    '''
    Adds Google Earth Engine Image Layer to a Folium Map.
    Args:
        self     (object):  folium map object to which the layer will be added.
        ee_image_object (ee.image.Image): Google Earth Engine image to be added to the folium map.
        vis_params (dict):  visualization parameters for the image.
            Must contain at least one 'bands' property:
            Exemple:{'bands':['SR_B4','SR_B3','SR_B2'],'max':.3}.
        name        (str):  name of the layer to be added.
    Retorn: None
    Example:
        # Defining Folium Map:
        map   = folium.Map(location=[-15.783333, -47.866667], zoom_start=12)
        # GEE Image:
        imagem = ee.Image('LANDSAT/LC08/C01/T1_SR/LC08_220071_20170805')
        # Visualization ParaMeters:
        vis    = {'bands':['B4','B3','B2'],'max':.3}
        # Adding Image to Map:
        map.add_ee_layer(image, vis,'Brasilia LandSat')
        # Map DisPlay:
        display(map)
    '''
    tiles =ee.Image(img).getMapId(params)
    folium.raster_layers.TileLayer(tiles  = tiles['tile_fetcher'].url_format,
                                   attr   ='Map Data &copy; <a href=https://earthengine.google.com/>Google Earth Engine</a>',
                                   name   = name,
                                   overlay= True,
                                   control= True).add_to(self)
# NDVI:
def normalize255(ndvi):
    '''
    Normalize NDVI Matrix to Band BetWeen 0 & 255.
    Arguments:
    ndvi - numpy matriz representing image NDVI
    Return:
    ndvi -- normalized NDVI matrix
    '''
    ndvi = ndvi+1
    ndvi =(ndvi/2)*255
    return ndvi.astype(np.uint8)    # Ensure it is explicitly converted to uint8 for image processing functions
def getNDVI(ds):
    '''
    Calculates Normalized Differentiation Vegetation Index (NDVI) for an image.
    Arguments:
    ds - gdal open data object representing the image
    Return:
    ndvi - numpy matrix representing image NDVI
    '''
   #red  = ds.GetRasterBand(4).ReadAsArray( )   # 4: RED
   #nir  = ds.GetRasterBand(5).ReadAsArray( )   # 5: NIR
   #ndvi = (nir - red)/(nir +  red)
   #ndvi = normalize255(ndvi)
   #return ndvi
    try:
        # Read Bands & Convert to float32 for Accurate Division:
        red=ds.GetRasterBand(4).ReadAsArray( ).astype(np.float32)   # 4: RED
        nir=ds.GetRasterBand(5).ReadAsArray( ).astype(np.float32)   # 5: NIR
        # Handle potential zero in denominator by adding small epsilon or by using np.divide with a 'where' clause
        # np.seterr is a simpler way for a specific function context:
        with np.errstate(divide='ignore', invalid='ignore'):ndvi=(nir-red)/(nir+red)
        # Handle NaN & Inf that might result from division by zero replace with 0 or specific no-data value:
        ndvi=np.nan_to_num(ndvi, nan=.0, posinf=.0, neginf=.0)      # Convert NaN/inf to 0
        # Normalize to 0-255 & Convert to uint8:
        ndvi=normalize255(ndvi)
        return ndvi
    except Exception as e:
        print(f'Error Calculating NDVI: {e}')
        # Return an array of zeros or raise the exception, depending on desired behavior:
        return np.zeros((ds.RasterYSize, ds.RasterXSize), dtype=np.uint8)
def getNDVImask(ndvi):
    '''
    Creates bounding mask vegetation area in NDVI matrix.
    Arguments:
    ndvi -  numpy matrix representing NDVI.
    Return:
    mask - binary matrix where 1 is vegetation.
    '''
   #_, mask = cv.threshold(ndvi, 170, 255, cv.THRESH_BINARY)
   #return mask
    # Ensure InPut is uint8, as cv.threshold expects this:
    if ndvi.dtype!=np.uint8:
        print('Warning: NDVI Mask InPut is not uint8. Converting…')
        ndvi=ndvi.astype(np.uint8)
    # ThresHold Chosen for Normalized NDVI (145 Corresponds to Approx. 0.137 Original NDVI)
    threshold, mask=cv.threshold(ndvi, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU) # 145 | 150 | 170 | 180 | 191.25 | 197.5
    return mask, round(threshold)
def forestPerCent(mask):
    '''
    Calculates   vegetation pixel percentage on a mask.
    Arguments:
         mask:       binary matrix where 1 represents vegetation.
    Return:
        percent: vegetation pixels percentage.
    '''
    c =cv.countNonZero(mask)            # Count Non-Zero Pixels (Vegetation is 255)
    t =mask.shape[0] * mask.shape[1]    # Total          Pixels
    if t==0:return          .0          # Avoid Division by Zero if Mask is Empty
    return round((c / t) * 100 ,  2)
def plotNDVI(path, save=None):
    '''
    Plots NDVI & vegetation mask showing vegetation percentage.
    Arguments:
         path:   path to image file.
    save_path:  (Optional) path to save the generated plot image (e.g.,'results/ndvi_plot_2025.png').
                  If None, plot is only displayed.
    '''
    # ds       =gdal.Open  (path)
    # if ds is None:
    #     print(f'Error: Could not open {path}!')
    #     return
    # ndvi     =getNDVI    (ds)
    # ndvi_mask=getNDVImask(ndvi)
    # year=path[33:37]
    # fig =plt.figure(figsize=(12, 4))
    # fig =plt.subplot(1, 2, 1)
    # plot=plt.imshow(ndvi     , cmap='gray')
    # plt.colorbar( )
    # plt.title(year+': '+'NDVI')
    # plt.yticks([]), plt.xticks([])
    # fig =plt.subplot(1, 2, 2)
    # plot=plt.imshow(ndvi_mask, cmap='Greens')
    # plt.title(str(forestPerCent(ndvi_mask))+'%')
    # plt.yticks([]), plt.xticks([])
    # plt.box(False)
    # plt.show  (  )
    ds =None # Initialize ds to None
    try:
        ds=gdal.Open(path)
        if ds is None:
            print(f'Error: Could not open GeoTIFF file at {path}. Please check path & file integrity.')
            return
        ndvi=getNDVI(ds)
        mask, threshold_value=getNDVImask(ndvi)
        # --- Robust Year Extraction ---
        # Get FileName:
        filename=os.path.basename(path)
        # Extract year, assuming format 'DFYYYY.tif'
        # Check if filename is long enough to safely slice
        if len(filename)>=6 and filename.startswith('DF'):year=filename[2:6]
        else:year        ='UnKnown Year' # FallBack if Format Does Not Match
        # ------------------------------
        # Creating Figure & SubPlots More Cleanly:
        fig, (ax1, ax2)  =plt.subplots(1, 2, figsize=(12, 4), tight_layout=True)
        # Plot NDVI:
        im1=ax1.imshow(ndvi, cmap='gray')
        fig.colorbar  (  im1,   ax=  ax1 )   # Attach ColorBar to ax1
        ax1.set_title(f'{year} NDVI {threshold_value} Otsu ThresHold')
        ax1.set_yticks([]), ax1.set_xticks([]) # Remove Ticks
        # Plot NDVI Mask:
        im2=ax2.imshow(mask, cmap='Greens') # Use 'Greens' ColorMap for Mask
        # No ColorBar needed for Binary Mask, unless want to show 0/255 scale
        fig.colorbar  (  im2,   ax=  ax2 )
        ax2.set_title(f' Vegetation Cover @ {forestPerCent(mask)}%')
        ax2.set_yticks([]), ax2.set_xticks([]) # Remove Ticks
        plt.grid        ('off')
        plt.box         (False) # Remove Outer Figure Box (if any)
        if save:
            # Create Directory if it Does not Exist:
            os.makedirs(os.path.dirname(save), exist_ok=True)
            # print(f'Plot saved to: {save_path}\n')
            # bbox_inches='tight' to prevent labels/titles from being cut off
            plt.savefig(save, bbox_inches='tight', dpi=300)
        plt.show        (     )
    except gdal.CPLE_BaseError as e:print(f'GDAL Error during processing "{path}": {e}')
    except           Exception as e:print(f'An unexpected error occurred during processing "{path}": {e}')
    finally:
        # Ensure DataSet is Closed by Closing GDAL DataSet:
        if ds is not None:ds=None
print('Functions Loaded.')
````

````python
# @title Saving GeoTiff Images
# Task Config:
taskConfig={'folder'       : 'LandSat',
            'scale'        :  30,
            'region'       :  geometry,
            'fileFormat'   : 'GeoTIFF',
            'maxPixels'    :  1e9,
            'formatOptions':{'cloudOptimized':True}}
task1=ee.batch.Export.image.toDrive(landsat00, description='DF2000', **taskConfig)
task2=ee.batch.Export.image.toDrive(landsat05, description='DF2005', **taskConfig)
task3=ee.batch.Export.image.toDrive(landsat10, description='DF2010', **taskConfig)
task4=ee.batch.Export.image.toDrive(landsat15, description='DF2015', **taskConfig)
task5=ee.batch.Export.image.toDrive(landsat20, description='DF2020', **taskConfig)
task6=ee.batch.Export.image.toDrive(landsat25, description='DF2025', **taskConfig)
def geeTasks(tasks, drive_folder='LandSat'):
    '''
    Checks if expected OutPut GeoTIFF file for each GEE task already exists
    in the specified Google Drive folder; if file is missing, task is started.
    Arguments:
        tasks      (list): list of ee.batch.Task objects (e.g.,[task1,task2,task3]).
        drive_folder(str): GEE OutPut Folder Name        (e.g.,'LandSat').
    '''
    # 1. Define Base Path for Google Drive:
    drive_path=os.path.join('/content/drive/MyDrive', drive_folder)
    # Ensure Local Directory Structure Exists for Checking Purposes:
    os.makedirs(drive_path, exist_ok=True)
    print('--- GEE Task Execution Check ---')
    tasks_to_start=[ ]
    for task in tasks:
        description=task.config['description']    # e.g.,'DF2015'
        # 2. Construct Full Expected File Path:
        expected_filename=f'{description}.tif'
        full_path=os.path.join(drive_path, expected_filename)
        # 3. Check for File Existence:
        if os.path.exists(full_path):print(f'✅ SKIPPING task {description}: File already exists at {full_path}')
        else:
            tasks_to_start.append(task)
            print(f'🔄 STARTING task {description}: File Not Found. Initiating Export…')
    # 4. Start Only Necessary Tasks:
    if tasks_to_start:
        for task in tasks_to_start:task.start( )
        print (f'\nStarted {len(tasks_to_start)} GEE export tasks.')
    else:print('\nAll required GeoTIFF files were found. No GEE tasks started.')
# Group Previously Defined Tasks
all_tasks=[task1, task2, task3, task4, task5, task6]
# Execute Check & Conditional Start:
geeTasks(all_tasks, drive_folder='LandSat')
# task1.start( )
# task2.start( )
# task3.start( )
# task4.start( )
# task5.start( )
# task6.start( )
# time.sleep(900)
# task1.status( )
# task2.status( )
# task3.status( )
# Asynchronous Task Polling:
def TasksPolling(tasks, minutes=15):
    '''
    Continuously checks the status of a list of GEE export tasks.
    Blocks execution until ALL tasks are 'COMPLETED','UNSUBMITTED', or 'FAILED'.
    Arguments:
        tasks (list): list of ee.batch.Task objects (e.g., [task1, task2, task3]).
        minutes(int): minutes to wait between status checks.
    '''
    seconds=minutes*60
    # Creates List  of Task Descriptions for Easy Tracking:
    descriptions=[t.config['description']for t in tasks]
    total_tasks = len(tasks)
    print(f'Checking Status for {total_tasks} tasks every {minutes} minutes:')
    while True:
        # Check Status of All Tasks in the List:
        statuses=[t.status( ) for t in tasks]
        SUCCESS =['COMPLETED','UNSUBMITTED']
        # Count the Number of Completed & Running/Ready Tasks:
        completed_count=sum(1 for s in statuses if s['state']in SUCCESS)
        failed_count   =sum(1 for s in statuses if s['state']=='FAILED')
        # Log Current Progress:
        print(f'\n--- {time.strftime("%Y.%m.%d %H:%M:%S")} ---')
        print(f'{completed_count} of {total_tasks} tasks done (Completed or UnSubmitted).')
        # Print Status for Each Task:
        for i, status in enumerate(statuses):
            state=status['state']
            desc =descriptions[i]
            if   state==     'FAILED':print(f'❌  FAILED: {desc}. Error: {status.get("error_message","Unknown Error")}')
            elif state=='UNSUBMITTED':print(f'✅ SKIPPED: {desc}')
            # else                     :print(f'[{state:9}]{desc}')
            else                     :print(f'{desc} {state}')
        # CONDITION TO EXIT THE LOOP:
        if completed_count==total_tasks:
            print('\nAll GEE tasks finalized (Completed or Skipped). Proceeding…\n')
            break
        # CONDITION TO HANDLE FAILURES:
        if failed_count>0:raise RuntimeError('One or more GEE tasks FAILED. Check logs for details.')
        # PAUSE – Wait Before Next Check:
        print(f'rechecking in {minutes} minutes…')
        time.sleep(seconds)
# EXECUTION
# Group Tasks Into a Single List:
export_tasks=[task1, task2, task3, task4, task5, task6]
# Call Function to Poll Tasks:
TasksPolling(export_tasks, minutes=5)
task6.status( )
````
